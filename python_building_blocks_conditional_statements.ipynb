```python
  # Simple Calculator Test1

a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
operator = input("Enter an operator (+, -, *, /): ")
print("Test1")
if operator == "+":
    result = a + b
    print("Result =", result)

elif operator == "-":
    result = a - b
    print("Result =", result)

elif operator == "*":
    result = a * b
    print("Result =", result)

elif operator == "/":
    if b != 0:
        result = a / b
        print("Result =", result)
    else:
        print("Error: Division by zero is not allowed.")

else:
    print("Invalid operator. Please use +, -, *, or /.")
print()#blank space between results

  # Simple Calculator Test2

a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
operator = input("Enter an operator (+, -, *, /): ")
print("Test2")
if operator == "+":
    result = a + b
    print("Result =", result)

elif operator == "-":
    result = a - b
    print("Result =", result)

elif operator == "*":
    result = a * b
    print("Result =", result)

elif operator == "/":
    if b != 0:
        result = a / b
        print("Result =", result)
    else:
        print("Error: Division by zero is not allowed.")

else:
    print("Invalid operator. Please use +, -, *, or /.")
print()#blank space between results

  # Simple Calculator Test3

a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))
operator = input("Enter an operator (+, -, *, /): ")
print("Test3")
if operator == "+":
    result = a + b
    print("Result =", result)

elif operator == "-":
    result = a - b
    print("Result =", result)

elif operator == "*":
    result = a * b
    print("Result =", result)

elif operator == "/":
    if b != 0:
        result = a / b
        print("Result =", result)
    else:
        print("Error: Division by zero is not allowed.")

else:
    print("Invalid operator. Please use +, -, *, or /.")
print()#blank space between results
```
